# E-Commerce Logistics Network Optimization
## Exploratory Analysis Notebook

This notebook explores the logistics network data and validates the shortest path results produced by `src/solution.py`.

## 1. Import Libraries

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

print('Libraries loaded successfully.')
```

## 2. Load and Explore the Dataset

```python
df = pd.read_csv('../data/network_data.csv')
print(f'Dataset shape: {df.shape}')
df.head(14)
```

```python
df[['travel_time_min', 'distance_km', 'cost_usd']].describe()
```

## 3. Build the Network Graph

```python
G = nx.DiGraph()
for _, row in df.iterrows():
    G.add_edge(row['source'], row['target'],
               travel_time_min=row['travel_time_min'],
               distance_km=row['distance_km'],
               cost_usd=row['cost_usd'])
print(f'Nodes: {G.number_of_nodes()}')
print(f'Edges: {G.number_of_edges()}')
```

## 4. Shortest Path Analysis

```python
source = 'Warehouse_Istanbul'
targets = ['Customer_Zone_A', 'Customer_Zone_B']

for target in targets:
    path = nx.shortest_path(G, source=source, target=target, weight='travel_time_min')
    time = nx.shortest_path_length(G, source=source, target=target, weight='travel_time_min')
    dist = sum(G[path[i]][path[i+1]]['distance_km'] for i in range(len(path)-1))
    cost = sum(G[path[i]][path[i+1]]['cost_usd'] for i in range(len(path)-1))
    print(f'Target: {target}')
    print(f'  Route: {" -> ".join(path)}')
    print(f'  Time: {time} min | Distance: {dist} km | Cost: ${cost}')
```
